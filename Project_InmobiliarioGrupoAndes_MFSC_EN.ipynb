# 📊 Project: Real Estate Commercial Analysis

The real estate sector needs to evaluate its **commercial performance** to understand its growth, profitability, and customer behavior.

The request is to build an **executive dashboard** that enables analysis of sales, customers, and properties to support data-driven strategic decisions.

# 📂 Dataset Description

🔹 **hecho_ventas_propiedades**  
Fact table with sales transactions (price, customer, property, channel, date).

🔹 **dim_clientes**  
Customer information and segmentation.

🔹 **dim_propiedades**  
Property characteristics (type, size, location, etc.).

🔹 **dim_fecha**  
Calendar table to be created as part of the modeling.

# 🎯 Questions the Dashboard Must Answer

**General Performance**

- What is the **total revenue** generated by the company?
- How many **properties have been sold**?
- What is the **average sale price**?
- What is the **total commission** generated by the company?

**Commercial Analysis**

- Which **property type generates the most revenue**?
- Which **customer segment buys the most properties**?
- Which sales channel **generates the most revenue**?


**Temporal Analysis**

- How do **sales evolve over time**?
- Is the business **growing year over year**?
- How is **current-year-to-date (YTD) performance** tracking?

**Customer Cohorts**

- Do customers **repurchase** after their first purchase?
- Which **customer cohorts generate more purchases/revenue over time**?

## 📝 Step 6: Executive Summary

**Key Findings**
- Total revenue for the period (Jan 2023–Dec 2024) was USD 6,012.5 million, generated by 8,500 sales.
- The property type generating the most revenue is Casa (House), with USD 2,240.5 M (although Departamento (Apartment) has the highest unit volume: 5,105 sales, reflecting a much lower average ticket: USD 361,856 vs. USD 964,086 for Casa).
- The city with the highest revenue is Mexico City (USD 3,242.2 M / 4,123 sales), slightly above Bogotá (USD 2,770.3 M / 4,377 sales) — Bogotá sells more units but at a lower average value.
- The most revenue-efficient sales channel is Corredor (Broker), with USD 4,380.0 M (73% of total revenue) versus USD 1,632.5 M from the Direct channel.

**Key Metrics**
- Total Revenue: USD 6,012,502,170
- Number of Sales: 8,500
- Average Ticket: USD 707,353
- Total Commission: USD 200,627,166

**Actionable Insights**
- The First-Time segment concentrates most of the revenue (USD 3,783.6 M, 63% of the total, with 5,302 sales), followed by Investor (USD 1,471.4 M) and High Net Worth (USD 757.5 M, the lowest-volume segment but with the highest value per transaction).
- Customers acquired in the 2023-02 cohort show the highest recurrence (93.0% of their customers repurchased).
- Sales show growth of +11.1% YoY (2023: USD 2,847.7 M → 2024: USD 3,164.8 M).

**Strategic Recommendations**
- **Prioritize the commercialization of Casa and Comercial (Commercial) property types**, which generate the highest average ticket (USD 964,086 and USD 1,797,098 respectively), optimizing the revenue mix without relying solely on Departamento volume.
- **Strengthen the Corredor (Broker) sales channel**, which concentrates 73% of revenue, through incentives or loyalty programs, given its greater revenue share, and evaluate how to scale the Direct channel to reduce total commission cost.
- **Implement retention strategies** to improve repurchase in recent cohorts, targeting "First-Time" customers specifically, since they are the largest segment and the one with the highest aggregate revenue; converting them into repeat buyers (following the 77% recurrence pattern already observed) has high incremental revenue potential.
- **Capitalize on the seasonality detected in YoY growth**, reinforcing commercial campaigns in lower-performing months to smooth cyclicality and sustain the +11% annual growth.
